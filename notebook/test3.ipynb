```python
# Import necessary libraries
import camelot
import pandas as pd
```

```python
# Define the extraction function
def extract_tables_from_pdf(file_path):
    """
    Extract useful financial tables (like income statements) from a PDF.
    We only scan page range 170-end to avoid unnecessary noise.
    """
    try:
        # Focus on financial statements pages
        tables = camelot.read_pdf(file_path, pages='130-end', flavor='stream')

        if tables.n == 0:
            print(f"No tables found in {file_path}")
            return []

        useful_tables = []
        keywords = ['total revenue', 'cost of sales', 'operating expenses', 'gross profit','earnings per share', 'net asset','number of shares','%','top twenty shareholders of the company']

        for i, table in enumerate(tables):  
            df = table.df
            # Flatten table text
            table_text = ' '.join(df.apply(lambda row: ' '.join(row), axis=1).tolist()).lower()

            if any(keyword in table_text for keyword in keywords):
                print(f"Selected table {i} (financial content detected)")
                useful_tables.append(df)
            else:
                print(f"Skipped table {i} (non-financial content)")

        if not useful_tables:
            print(f"No financial tables found in {file_path}")

    
        return useful_tables

    except Exception as e:
        print(f"Error extracting tables from {file_path}: {str(e)}")
        return [] 
```

```python
# Now, in your Jupyter Notebook:
# Extract the tables from the PDF (replace 'your_pdf_file.pdf' with your actual file path)
file_path = '508_1653300092463.pdf'
tables = extract_tables_from_pdf(file_path)
```

```python
# Initialize a list to store extracted data
data_for_csv = []

for i, table in enumerate(tables):
    # Convert all cells to lowercase and strip spaces
    table_lower = table.astype(str).map(lambda x: x.strip().lower())

    # Flag to track if a matching row was found
    found = False

    # Check each row
    for _, row in table_lower.iterrows():
        row_list = row.tolist()
        number_of_count = sum("number of" in cell for cell in row_list)
        percent_count = sum("%" in cell for cell in row_list)
    
        if number_of_count == 2 and percent_count == 2:
            print("Row has at least two 'number of' and two '%' entries")
            display(table)

            # Extract data starting from row 1 (assuming row 0 is header)
            share_holder_names = table.iloc[1:, 0].astype(str).str.strip().tolist()
            share_percentage = table.iloc[1:, 2].astype(str).str.strip().tolist()
            year = [2020] * len(share_holder_names)

            for name, percentage, yr in zip(share_holder_names, share_percentage, year):
                data_for_csv.append([yr, name, percentage])

            print(year)
            print(share_holder_names)
            print(share_percentage)
        

            found = True
            break  # stop checking more rows in the same table

# Save extracted data to CSV
df_csv = pd.DataFrame(data_for_csv, columns=["Year", "Shareholder Name", "Share Percentage"])
df_csv.to_csv("shareholder_data.csv", index=False, mode='a', header=False)

print("Data has been appended to 'shareholder_data.csv'")
```
